```python
fifa.drop(columns=["home_team_goalkeeper_score","away_team_goalkeeper_score","home_team_mean_defense_score","home_team_mean_offense_score","home_team_mean_midfield_score",
                   "away_team_mean_defense_score","away_team_mean_offense_score","away_team_mean_midfield_score"],inplace=True)
fifa.head(10)
```

```python
fifa["date"]=pd.to_datetime(fifa["date"])
```

```python
fifa.info()
```

```python
fifa["year"]=fifa["date"].dt.year
fifa.head()
```

```python
fifa["total_goals"] = fifa["home_team_score"] + fifa["away_team_score"]
```

```python
fifa[["home_team_score", "away_team_score", "total_goals"]].head()
```

```python

home_teams_match = fifa["home_team"].value_counts()
away_team_match = fifa["away_team"].value_counts()

total_matches = home_teams_match.add(away_team_match, fill_value=0)

total_matches = total_matches.sort_values(ascending=False)


top10_teams = total_matches.head(10)

plt.figure(figsize=(10,6))

top10_teams.plot(
    kind="barh",
    color="royalblue",
    edgecolor="black"
)

plt.title("Top 10 Teams by Total International Matches", fontsize=14)
plt.xlabel("Number of Matches")
plt.ylabel("Teams")

plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
home_goals = fifa.groupby("home_team")["home_team_score"].sum()
away_goals = fifa.groupby("away_team")["away_team_score"].sum()
total_goals = home_goals.add(away_goals, fill_value=0)

top10_goals = total_goals.sort_values(ascending=False).head(10)
top10_goals.plot(kind="bar", color="royalblue")

plt.title("Top 10 Goal Scoring Teams")
plt.xlabel("Teams")
plt.ylabel("Goals")

plt.show()
```

```python
tournament = fifa["tournament"].value_counts()
top10_tournament = tournament.head(10)

plt.figure(figsize=(12,6))

top10_tournament.plot(kind="bar", color="royalblue")

plt.title("Top 10 Tournaments by Number of Matches")
plt.xlabel("Tournament")
plt.ylabel("Matches")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

```python
fifa["home_team_result"].value_counts()
```

```python
result = fifa["home_team_result"].value_counts()
result.plot(kind="pie", autopct="%1.1f%%", colors=["#A1C9F4", "#FFB482", "#8DE5A1"])

plt.title("Home Win vs Away Win vs Draw")
plt.ylabel("")

plt.show()
```

```python
goals_year = fifa.groupby("year")["total_goals"].sum()
goals_year.head()
```

```python
avg_goals = fifa.groupby("year")["total_goals"].mean()

avg_goals.plot(kind="line", marker="o")

plt.title("Average Goals per Match by Year")
plt.xlabel("Year")
plt.ylabel("Average Goals")
plt.grid(True)

plt.show()
```

```python
home_goals = fifa.groupby("year")["home_team_score"].sum()
away_goals = fifa.groupby("year")["away_team_score"].sum()


plt.figure(figsize=(12,6))

plt.plot(home_goals.index, home_goals.values,
         marker="o", label="Home Goals")

plt.plot(away_goals.index, away_goals.values,
         marker="s", label="Away Goals")

plt.title("Home Goals vs Away Goals Over Years")
plt.xlabel("Year")
plt.ylabel("Total Goals")

plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
matches_per_year = fifa["year"].value_counts().sort_index()

plt.figure(figsize=(12,6))

plt.plot(matches_per_year.index,
         matches_per_year.values,
         marker="o",
         linewidth=2,
         color="royalblue")

plt.title("International Matches Played Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Matches")
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
continent = fifa["home_team_continent"].value_counts()

plt.figure(figsize=(8,6))

continent.plot(
    kind="bar",
    color="steelblue",
    edgecolor="black"
)

plt.title("Home Team Matches by Continent", fontsize=15, fontweight="bold")
plt.xlabel("Continent")
plt.ylabel("Number of Home Matches")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
home_win = fifa[fifa["home_team_score"] > fifa["away_team_score"]]["home_team_continent"].value_counts()

plt.figure(figsize=(8,6))

home_win.plot(
    kind="bar",
    color="steelblue",
    edgecolor="black"
)

plt.title("Home Wins by Continent")
plt.xlabel("Continent")
plt.ylabel("Number of Home Wins")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```
